## Module 13 — HTTP & REST APIs
### 13.1 - 13.6 HTTP Requests, Headers, Status Codes & Calling Web APIs
Perform HTTP requests to fetch external data and interact with REST APIs.


```python
# === PREDEFINED SOLUTION PROGRAM (IN COMMENTS) ===
# Uncomment (select lines & press Ctrl+/) or use as reference:
# --- Module 13 Practice: HTTP & REST API Requests ---
import urllib.request
import json

def fetch_public_api_data():
    url = "https://api.github.com/12zen"
    headers = {"User-Agent": "Python-Agentic-AI-Book"}

    req = urllib.request.Request(url, headers=headers)
    try:
        with urllib.request.urlopen(req) as response:
            status_code = response.getcode()
            content = response.read().decode("utf-8")
            print(f"HTTP Status: {status_code}")
            print(f"GitHub Zen Quote: '{content}'")
    except Exception as e:
        print("HTTP Request Error:", e)

fetch_public_api_data()

```

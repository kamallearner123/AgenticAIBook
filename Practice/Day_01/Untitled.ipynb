````python
# Program 5.0-1: Specialized Code Review Agent Persona
from openai import OpenAI
OPENAI_KEY = '...'
client = OpenAI()

system_persona = """
You are a Senior Principal Systems Architect specializing in Rust and high-performance networking.

MISSION:
Review provided code snippets for:
1. Memory safety & allocation inefficiencies (e.g., unnecessary cloning, string allocations in hot paths).
2. Concurrency hazards (race conditions, deadlock risks).
3. Idiomatic Rust design principles.

TONE & STYLE:
- Direct, clinical, and constructive.
- No pleasantries ("Hello", "Hope you are doing well", "Certainly!").
- Reference Rust compiler rules or RFC specifications where applicable.

GUARDRAILS:
- Do not evaluate non-systems programming languages.
- If the snippet is incomplete or missing context, flag the specific missing types.
"""

user_code = """
pub fn process_sensor_stream(data: Vec) -> String {
    let mut accumulator = String::new();
    for byte in data {
        // Repeated string allocation in loop
        accumulator.push_str(&byte.to_string());
    }
    accumulator
}
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0.0,
    messages=[
        {"role": "system", "content": system_persona},
        {"role": "user", "content": f"Review this code:\n```rust\n{user_code}\n```"}
    ]
)

print(response.choices[0].message.content)
````
